```python
import pandas as pd

# Load all three datasets
df_in = pd.read_csv('/INvideos.csv')
df_us = pd.read_csv('/USvideos.csv')
df_gb = pd.read_csv('/GBvideos.csv')

# Display dataset information
datasets = {
    "India": df_in,
    "USA": df_us,
    "Great Britain": df_gb
}

for country, data in datasets.items():
    print(f"\n{country}")
    print("Rows:", len(data))
    print("Columns:", len(data.columns))
    print("Missing values:", data.isnull().sum().sum())
    print("Exact duplicates:", data.duplicated().sum())
```

## 10. Exception Handling — Handling Invalid Video Data

```python
def safe_convert_views(value):
    try:
        return int(value)

    except (ValueError, TypeError):
        return 0
```

```python
conversion_errors = 0

for video in video_records:
    original_views = video["views"]
    converted_views = safe_convert_views(original_views)

    if converted_views == 0 and original_views != 0:
        conversion_errors += 1

print("Records processed:", len(video_records))
print("Conversion errors:", conversion_errors)
```

## 11. Date Parsing — YouTube Video Dates

```python
from datetime import datetime

def parse_trending_date(date_value):
    try:
        return datetime.strptime(
            date_value,
            "%y.%d.%m"
        )

    except (ValueError, TypeError):
        return None
```

```python
date_errors = 0

for video in video_records:
    parsed_date = parse_trending_date(
        video["trending_date"]
    )

    video["parsed_trending_date"] = parsed_date

    if parsed_date is None:
        date_errors += 1

print("Total records processed:", len(video_records))
print("Invalid dates:", date_errors)

print("\nFirst parsed date:")
print(video_records[0]["parsed_trending_date"])
```

## 12. YouTube Video Title Feature Extraction

```python
import re

def extract_title_features(title):

    # Handle missing or invalid titles
    if not isinstance(title, str):
        title = ""

    # Extract words
    words = re.findall(r"\b\w+\b", title)

    # Extract hashtags
    hashtags = re.findall(r"#\w+", title)

    # Create feature dictionary
    features = {
        "title_length": len(title),
        "word_count": len(words),
        "hashtag_count": len(hashtags),
        "hashtags": hashtags,
        "is_uppercase": title.isupper(),
        "contains_question": "?" in title,
        "contains_exclamation": "!" in title
    }

    return features
```

```python
sample_title = "TOP 10 AI TOOLS YOU MUST TRY! #AI #TECH"

sample_features = extract_title_features(sample_title)

print("Sample Title:", sample_title)
print("\nExtracted Features:")

for feature, value in sample_features.items():
    print(feature, ":", value)
```

```python
for video in video_records[:10000]:

    title_features = extract_title_features(
        video["title"]
    )

    video.update(title_features)

print("First 10,000 video titles processed successfully!")
```

```python
for video in video_records[10000:]:

    title_features = extract_title_features(
        video["title"]
    )

    video.update(title_features)

print("All video titles processed successfully!")
print("Total records processed:", len(video_records))
```

## 13. Rule-Based Video Classification

```python
def classify_content(title):

    title = str(title).lower()

    keywords = {
        "Education": ["learn", "tutorial", "course", "python", "programming", "education"],
        "Gaming": ["gaming", "gameplay", "minecraft", "gamer", "pubg"],
        "Entertainment": ["movie", "trailer", "comedy", "funny", "music"],
        "Sports": ["cricket", "football", "soccer", "basketball", "highlights"],
        "News": ["news", "breaking", "politics", "election"]
    }

    for category, words in keywords.items():

        for word in words:

            if word in title:
                return category

    return "Other"
```

```python
for video in video_records:

    video["content_category"] = classify_content(
        video["title"]
    )

print("Video classification completed!")
print("Total records:", len(video_records))
```

```python
from collections import Counter

category_counts = Counter(
    video["content_category"]
    for video in video_records
)

print("Content Classification Summary:\n")

for category, count in category_counts.items():
    print(category, ":", count)
```

```python
for video in video_records[:10]:

    print(
        "Title:", video["title"],
        "| Classification:", video["content_category"]
    )
```

```python
def process_video_record(record):

    processed = record.copy()

    # Convert views safely
    views = safe_convert_views(processed.get("views"))
    processed["views_numeric"] = views

    # Parse trending date
    processed["parsed_trending_date"] = parse_trending_date(
        processed.get("trending_date")
    )

    # Extract title features
    title_features = extract_title_features(
        processed.get("title")
    )
    processed.update(title_features)

    # Classify video
    processed["popularity_category"] = classify_video(views)
    processed["content_category"] = classify_content(
        processed.get("title")
    )

    return processed
```

```python
processed_video_records = [
    process_video_record(record)
    for record in video_records
]

print("Total processed records:", len(processed_video_records))
print("Original records:", len(video_records))
```
